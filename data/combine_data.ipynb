```python
import pandas as pd

def combine_datasets():
    print("Memulai proses penggabungan dataset...")

    # 1. Membaca kedua file CSV
    # Pastikan file CSV berada di satu folder yang sama dengan skrip ini
    try:
        df_train = pd.read_csv('train.csv')
        df_insta = pd.read_csv('Instagram_fake_profile_dataset.csv')
        print(f"Jumlah baris train.csv: {len(df_train)}")
        print(f"Jumlah baris Instagram_fake_profile_dataset.csv: {len(df_insta)}")
    except FileNotFoundError as e:
        print(f"Error: File tidak ditemukan. Pastikan nama file sesuai. Detail: {e}")
        return

    # 2. Menggabungkan kedua dataset (menumpuk baris ke bawah)
    df_combined = pd.concat([df_train, df_insta], ignore_index=True)
    print(f"\nTotal baris setelah digabung: {len(df_combined)}")

    # 3. Menghapus data duplikat (opsional tapi sangat direkomendasikan)
    # Sangat mungkin ada akun yang sama muncul di kedua file tersebut
    df_cleaned = df_combined.drop_duplicates()
    
    duplikat_dihapus = len(df_combined) - len(df_cleaned)
    print(f"Ditemukan dan dihapus {duplikat_dihapus} baris data duplikat.")
    print(f"Total baris dataset bersih final: {len(df_cleaned)}")

    # 4. Menyimpan dataset baru
    output_filename = 'master_dataset_cleaned.csv'
    df_cleaned.to_csv(output_filename, index=False)
    print(f"\nBerhasil! Dataset utama telah disimpan sebagai '{output_filename}'")

if __name__ == "__main__":
    combine_datasets()
```
